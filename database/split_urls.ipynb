```python
import pandas as pd

PHISH_ONLY = "한국인터넷진흥원_피싱사이트 URL_20231231.csv"
BENIGN_ONLY = "recent-top-1m.csv"
MIXED = "malicious_phish_benign_phishing.csv"

# 피싱파일
kisa = pd.read_csv(PHISH_ONLY, encoding="utf-8-sig")  
phish_from_kisa = kisa[["홈페이지주소"]].rename(columns={"홈페이지주소": "url"})

# 정상파일
top1m = pd.read_csv(BENIGN_ONLY, header=None)
benign_from_top1m = top1m.iloc[:, [1]].rename(columns={1: "url"})

# 믹스파일
mixed = pd.read_csv(MIXED)[["url", "type"]]
mixed["type"] = mixed["type"].astype(str).str.lower().str.strip()

phish_from_mixed = mixed[mixed["type"] == "phishing"][["url"]]
benign_from_mixed = mixed[mixed["type"] == "benign"][["url"]]

# 합치기
phishing_all = pd.concat([phish_from_kisa, phish_from_mixed], ignore_index=True)
benign_all = pd.concat([benign_from_top1m, benign_from_mixed], ignore_index=True)

# 파일 저장
phishing_all.to_csv("phishing_urls.csv", index=False, header=False)
benign_all.to_csv("benign_urls.csv", index=False, header=False)
```
