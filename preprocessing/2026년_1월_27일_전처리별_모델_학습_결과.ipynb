# Recall > Precision > F1 > Accuracy

- Recall: 실제 피싱을 얼마나 놓치지 않고 탐지하는지 (보안 위험 통제 지표)

- Precision: 피싱으로 판단한 것 중 실제 피싱의 비율 (오탐/운영 비용 지표)

- Accuracy: 전체 예측 중 정답 비율 (불균형 데이터에서는 보조 지표)

- F1-score: Recall과 Precision의 균형을 나타내는 종합 성능 지표
